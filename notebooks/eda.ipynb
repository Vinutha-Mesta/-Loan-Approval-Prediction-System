# Loan Approval Prediction — Exploratory Data Analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='darkgrid', palette='muted')
df = pd.read_csv('../data/loan_data.csv')
print(df.shape)
df.head()
```

```python
# Basic info
df.info()
df.describe()
```

```python
# Target distribution
ax = df['Loan_Status'].value_counts().plot(kind='bar', color=['#ef4444','#10b981'], edgecolor='white')
ax.set_xticklabels(['Rejected','Approved'], rotation=0)
plt.title('Loan Status Distribution')
plt.tight_layout(); plt.show()
```

```python
# Credit History vs Approval
pd.crosstab(df['Credit_History'], df['Loan_Status']).plot(kind='bar', color=['#ef4444','#10b981'], edgecolor='white')
plt.title('Credit History vs Loan Approval')
plt.xlabel('Credit History (0=Bad, 1=Good)')
plt.tight_layout(); plt.show()
```

```python
# Income distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df[df['Loan_Status']==1]['ApplicantIncome'].hist(ax=axes[0], bins=30, color='#10b981', alpha=.8)
axes[0].set_title('Income — Approved')
df[df['Loan_Status']==0]['ApplicantIncome'].hist(ax=axes[1], bins=30, color='#ef4444', alpha=.8)
axes[1].set_title('Income — Rejected')
plt.tight_layout(); plt.show()
```

```python
# Correlation heatmap (numeric)
num_df = df.select_dtypes(include=np.number)
plt.figure(figsize=(8,6))
sns.heatmap(num_df.corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=.5)
plt.title('Feature Correlation Matrix')
plt.tight_layout(); plt.show()
```
